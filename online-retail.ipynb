```python
# Import Libraries
import pandas as pd
import numpy as np
```

```python
# Reading the dataset
df = pd.read_csv('online_retail.csv', encoding='latin1')
df.head()
```

```python
# Reading the data information
df.info()
```

```python
# Handling missing values
# Drop rows with missing CustomerID (as it is crucial for analysis)
df = df.dropna(subset=['CustomerID'])
df.info()
```

```python
# Fill missing Description column with 'Unknown' values
df['Description'] = df['Description'].fillna('Unknown')
```

```python
# Convert InvoiceDate Dtype to datetime
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df.info()
```

```python
duplicates = df.duplicated().sum()
print(f"\nNumber of duplicate rows: {duplicates}")
```

```python
# Identify duplicate rows
duplicate_rows = df[df.duplicated(keep=False)]
duplicate_rows
```

```python
# Remove duplicates
df = df.drop_duplicates()
```

```python
# Remove rows where Quantity is negative
df = df[df['Quantity'] > 0]
df.info()
```

```python
# Remove rows where UnitPrice is zero or negative
df = df[df['UnitPrice'] > 0]
df.info()
```

```python
# Create a TotalPrice column
df['TotalPrice'] = df['Quantity'] * df['UnitPrice']

# Extract date-related features
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['Day'] = df['InvoiceDate'].dt.day
df['Weekday'] = df['InvoiceDate'].dt.weekday
df['Hour'] = df['InvoiceDate'].dt.hour
df
```

```python
# Create a Revenue per Customer metric
customer_revenue = df.groupby('CustomerID')['TotalPrice'].sum().reset_index()
customer_revenue.rename(columns={'TotalPrice': 'TotalRevenue'}, inplace=True)
```

```python
# Convert CustomerID to integer
df['CustomerID'] = df['CustomerID'].astype(int)
```

```python
# Merge revenue data back to the main dataframe
#df = df.merge(customer_revenue, on='CustomerID', how='left')

# Display the cleaned dataset
print(df.info())
df
```

```python
# Save the cleaned data
df.to_csv("online_retail_cleaned.csv", index=False)
```
